```python
df_train['total_nutrients'] = df_train[['carbs_amount','protein_amount','fat_amount']].sum(axis=1)
df_test['total_nutrients'] = df_test[['carbs_amount','protein_amount','fat_amount']].sum(axis=1)
```

```python
df_train.head()
```

```python
df_train.drop(['carbs_amount','protein_amount','fat_amount'],axis=1,inplace=True)
df_test.drop(['carbs_amount','protein_amount','fat_amount'],axis=1,inplace=True)
```

```python
df_train['total_nutrients']=np.log1p(df_train['total_nutrients'])
df_test['total_nutrients']=np.log1p(df_test['total_nutrients'])
```

```python
plt.figure(figsize=(25,6))
df_train.boxplot()
```

```python
df_train['heart_rate_range'] = (df_train['heart_rate_max'] - df_train['heart_rate_avg'])
```

```python
df_test['heart_rate_range'] = (df_test['heart_rate_max'] - df_test['heart_rate_avg'])
```

```python
df_train.head()
```

```python
df_train.drop(['heart_rate_max','heart_rate_avg'],inplace=True,axis=1)
df_test.drop(['heart_rate_max','heart_rate_avg'],inplace=True,axis=1)
```

```python
df_train.head()
```

```python
df_train['sodium_density'] = (df_train['sodium_content_mg']/df_train['portion_size'])
df_test['sodium_density'] = (df_test['sodium_content_mg']/df_test['portion_size'])
```

```python
df_train.head()
```

```python
df_train.drop(['sodium_content_mg','portion_size'],axis=1,inplace=True)
df_test.drop(['sodium_content_mg','portion_size'],axis=1,inplace=True)
```

```python
df_train.drop('heart_rate_rest',axis=1,inplace=True)
df_test.drop('heart_rate_rest',axis=1,inplace=True)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
from sklearn.model_selection import train_test_split

X = df_train.drop('lean_body_mass',axis=1)
y = df_train['lean_body_mass']
```

```python
X_train,X_val,y_train,y_val = train_test_split(X,y,test_size=0.30,random_state=42)
```

```python
X_test = df_test.drop('lean_body_mass',axis=1)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LinearRegression

regression = LinearRegression()
regression.fit(X_train_scaled,y_train)
```

```python
y_pred=regression.predict(X_val_scaled)
```

```python
from sklearn.metrics import r2_score

print("LinearRegression :")
print(r2_score(y_val,y_pred))
```

```python
from xgboost import XGBRegressor

xgb = XGBRegressor(random_state=42)
xgb.fit(X_train_scaled,y_train)
```

```python
y_pred_xgb = xgb.predict(X_val_scaled)
```

```python
print(r2_score(y_val,y_pred_xgb))
```

```python
from sklearn.ensemble import RandomForestRegressor
forest = RandomForestRegressor(random_state=42)
forest.fit(X_train_scaled,y_train)
y_pred_forest = forest.predict(X_val_scaled)
print(r2_score(y_val,y_pred_forest))
```

```python
from sklearn.model_selection import cross_val_score

linear_cv = cross_val_score(
    regression,
    X_train_scaled,
    y_train,
    cv=5,
    scoring='r2'
)
```

```python
print(linear_cv.mean())
```

```python
xgb_cv = cross_val_score(
    xgb,
    X_train_scaled,
    y_train,
    cv=5,
    scoring='r2'
)
```

```python
print(xgb_cv.mean())
```

```python
param_grid = {
    'max_depth':[2,3,5],
    'learning_rate':[0.01,0.05,0.1],
    'n_estimators' :[50,100]
}
```

```python
from sklearn.model_selection import GridSearchCV

grid_search = GridSearchCV(
    estimator=xgb,
    param_grid=param_grid,
    cv=5,
    scoring='r2',
    n_jobs=-1
)
```

```python
grid_search.fit(X_train_scaled,y_train)
```

```python
print(grid_search.best_params_)
```

```python
print(grid_search.best_score_)
```

```python
best_xgb = grid_search.best_estimator_
best_xgb
```

```python
print(xgb.feature_importances_)
```

```python
importance_df = pd.DataFrame({
    'Feature' : X.columns,
    'Importance' : xgb.feature_importances_
})

importance_df = importance_df.sort_values(
    by='Importance',
    ascending=False
)

print(importance_df)
```

```python
from sklearn.model_selection import RandomizedSearchCV

param_dist = {
    'n_estimators': [50, 100, 200, 300],
    'max_depth': [2, 3, 4, 5, 6],
    'learning_rate': [0.001, 0.01, 0.03, 0.05, 0.1],
    'subsample': [0.6, 0.8, 1.0],
    'colsample_bytree': [0.6, 0.8, 1.0],
    'min_child_weight': [1, 3, 5]
}

random_search = RandomizedSearchCV(
    estimator=xgb,
    param_distributions=param_dist,
    n_iter = 30,
    scoring='r2',
    cv=5,
    verbose=2,
    random_state=42,
    n_jobs=-1
)
```

```python
random_search.fit(X_train_scaled,y_train)
```

```python
print(random_search.best_params_)
```

```python
print(random_search.best_score_)
```

```python
X_scaled = scaler.fit_transform(X)
X_test_scaled = scaler.transform(X_test)
```

```python
best_model = random_search.best_estimator_
```

```python
best_model.fit(X_scaled,y)
```

```python
final_pred = best_model.predict(X_test_scaled)
```

```python
df_test['lean_body_mass'] = final_pred
```
